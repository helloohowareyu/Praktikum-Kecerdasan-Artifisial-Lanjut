```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
import warnings
warnings.filterwarnings('ignore', category=DeprecationWarning)
```

```python
# Mensimulasikan dataset ritel
data = {
    'TransactionID': [1, 1, 2, 2, 3, 3, 3, 4, 4, 4],
    'Item': ['A', 'B', 'A', 'B', 'A', 'B', 'C', 'A', 'B', 'C']
}

df = pd.DataFrame(data)
df
```

```python
# Melakukan one-hot encoding untuk mempersiapkan data untuk analisis market basket.

basket = pd.crosstab(index=df['TransactionID'], columns=df['Item']).map(lambda x: 1 if x > 0 else 0)
basket
```

```python
# Menerapkan algoritma apriori untuk menemukan himpunan item yang sering muncul.
frequent_itemsets = apriori(basket, min_support=0.2, use_colnames=True)
frequent_itemsets
```

```python
# Menghasilkan aturan asosiasi
rules = association_rules(frequent_itemsets, metric='lift', min_threshold=1)
rules
```

Kolom yang relevan untuk diidentifikasi adalah InvoiceNO yang digunakan sebagai kunci unik transaksi untuk mengelompokkan item yang dibeli bersamaan (basket) dan Description sebagai variabel item yang akan dianalisis hubungannya dalam aturan asosiasi

Karena jumlah entri yang sangat besar (sektiar 541 ribu), maka akan dilakukan pemotongan jumlah entri

Alih-alih memotong data secara acak (yang bisa merusak pola belanja), cara yang saya gunakan adalah mengambil entri dari satu negara saja

Terlihat mayoritas data didapat dari negara UK dan entrinya adalah yang paling banyak, untuk itu kita gunakan negara lain dengan jumlah terbanyak kedua yaitu Jerman dengan jumlah entrinya adalah 603 yang cukup untuk digunakan di tugas kali ini.

```python
# Filter dataset untuk negara 'Germany'
df_germany = df[df['Country'] == 'Germany'].copy()
```

Selanjutnya kita akan mulai dengan membersihkan entri dengan menghapus transaksi dengan InvoiceNo yang kosong atau Nan dan menghapus transaksi yang dibatalkan (dimulai dengan 'C')

```python
df_germany.dropna(subset=['InvoiceNo', 'Description'], inplace=True)

df_germany = df_germany[~df_germany['InvoiceNo'].astype(str).str.contains('C')]
```

Selanjutnya adalah menghapus entri dengan kuantitas kurang dari atau sama dengan 0, yang biasanya adalah return barang. Selain itu kita akan bersihkan spasi berlebih di deskripsi produk

```python
df_germany = df_germany[df_germany['Quantity'] > 0]

df_germany['Description'] = df_germany['Description'].str.strip()
```

```python
df_germany.head(5)
```

### Soal 3
**Persiapan Data untuk Association Rule**

Di persiapan data, kita akan ubah data transaksi menjadi matriks dimana baris adalah ID transaksi dan kolom adalah item menggunakan pd.crosstab dan applymap untuk mengubah nilai (kuantitas) menjadi 1 jika barang tersebut dibeli dan 0 jika tidak

```python
basket = pd.crosstab(index=df_germany['InvoiceNo'], columns=df_germany['Description']).map(lambda x: 1 if x > 0 else 0)
basket
```

### Soal 4
**Penerapan Algoritma**

Kita akan terapkan algoritma apriori untuk mencari kombinasi item yang paling sering muncul sesering nilai `min_support`.

`min_support` yang kita gunakan kali ini adalah 0.02, artinya item harus muncul minimal di 2% dari seluruh transaksi.

```python
frequent_itemsets = apriori(basket, min_support=0.02, use_colnames=True)
frequent_itemsets
```

### Soal 5
**Evaluasi Aturan Asosiasi**

Setelah menerapkan apriori, kita akan menghasilkan aturan asosiasi dari `frequent_itemsets`

```python
rules = association_rules(frequent_itemsets, metric='lift', min_threshold=1)
```

Selanjutnya kita akan ambil 3 aturan terbaik berdasarkan nilai lift

```python
top_rules = rules.nlargest(3, 'lift')
top_rules
```

Dari hasil penerapan rules kita dapatkan:
1. Aturan [DOLLY GIRL CHILDRENS CUP] $\rightarrow$ [DOLLY GIRL CHILDRENS BOWL]: Memiliki nilai lift sebesar 38.08, yang menunjukkan korelasi positif yang kuat antara kedua produk tersebut. Dengan tingkat confidence sebesar 1
2. Aturan [DOLLY GIRL CHILDRENS BOWL] $\rightarrow$ [DOLLY GIRL CHILDRENS CUP]: Memiliki nilai lift sebesar 38.08, yang menunjukkan korelasi positif yang kuat antara kedua produk tersebut. Dengan tingkat confidence sebesar 0.9
3. Aturan [DOLLY GIRL CHILDRENS CUP, POSTAGE] $\rightarrow$ [DOLLY GIRL CHILDRENS BOWL]: Memiliki nilai lift sebesar 38.08, yang menunjukkan korelasi positif yang kuat antara kedua produk tersebut. Dengan tingkat confidence sebesar 1
